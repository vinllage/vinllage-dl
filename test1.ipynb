```python
import requests
from bs4 import BeautifulSoup as bs

url = "https://www.me.go.kr/mamo/web/index.do?menuId=631"

response = requests.get(url)
html_text = response.text

soup = bs(html_text, 'html.parser')
```

```python
keywords = ['환경', '수도권']
link_tags = soup.select(".brd_body .title a")
items = []
for tag in link_tags:
    title = tag.get_text().strip() # 게시글 제목 
    result = all(list(map(lambda x: x in title, keywords)))
    link = tag.attrs['href']
    if not result or not link: continue
  
    link = f"https://www.me.go.kr{link}"
    res = requests.get(link)
    _soup = bs(res.text, "html.parser")

    # 등록일
    el = _soup.select_one(".board_view .createDate")
    date = el.get_text().strip() if el else None
    
    # 게시글 내용 
    els = _soup.select(".board_view .board_con p")
    content = "\n".join([el.get_text() for el in els])
 
    items.append({"date": date, "link": link, "title": title, "content": content})
 
items
```
